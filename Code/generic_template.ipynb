```python
file_path = "../Datasets/GOLD_XYZ_OSC.0001_1024.hdf5"

idx = 0  # change this to view a different signal

with h5py.File(file_path, "r") as f:
    x = f["X"][idx]   # shape: (1024, 2)
    y = f["Y"][idx]   # shape: (24,)
    z = f["Z"][idx]   # shape: (1,)

I = x[:, 0]
Q = x[:, 1]

mod_class = np.argmax(y)
snr = z[0]

print("Signal shape:", x.shape)
print("Modulation class index:", mod_class)
print("SNR:", snr)

plt.figure(figsize=(12, 4))
plt.plot(I, label="I")
plt.plot(Q, label="Q")
plt.xlabel("Sample index")
plt.ylabel("Amplitude")
plt.title(f"I/Q Time-Domain Signal | Class={mod_class}, SNR={snr}")
plt.legend()
plt.grid(True)
plt.show()
```
